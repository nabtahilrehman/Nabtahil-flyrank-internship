## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
## 5. Why ML Beats a Fixed Rule Here

"""A fixed rule would require defining exact thresholds for multiple variables such as battery level, motor current, speed, distance, and tilt angle. However, robot behavior is influenced by combinations of these factors rather than a single measurement.

For example, two robots may have the same battery level but behave differently because of differences in speed, motor current, or tilt angle. These complex relationships make it difficult to create reliable if-then rules.

Machine learning clustering can automatically identify natural groups and hidden patterns in the data, allowing similar robot task executions to be grouped together without manually defining rules for every possible situation."""
```
