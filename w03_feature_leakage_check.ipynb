## 1. Build the feature vector

*Code that actually builds it — engineered features, categorical handling, fills.*

```python
## Feature Vector

"""The feature vector is designed to represent the search opportunity of a page.

I selected features that are available before making a recommendation and that may help explain visibility and traffic potential.

The feature vector combines search demand, rankings, engagement metrics, and content characteristics.

Missing values are filled with safe defaults and categorical variables are converted into numeric representations using one-hot encoding."""
```

```python
import pandas as pd
import os

# Define the URL for the raw CSV file on GitHub
github_raw_url = "https://raw.githubusercontent.com/nabtahilrehman/Nabtahil-flyrank-internship/main/data/raw/content_refresh_anonymized.csv"
local_file_path = "content_refresh_anonymized.csv" # Download to current working directory

# Check if the file exists locally, if not, download it
if not os.path.exists(local_file_path):
    print(f"Downloading {local_file_path} from GitHub...")
    !wget {github_raw_url} -O {local_file_path}
    print("Download complete.")
else:
    print(f"{local_file_path} already exists. Skipping download.")

df = pd.read_csv(local_file_path)

features = [
    "search_volume",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "engagement_rate",
    "scroll_rate",
    "content_age_days",
    "days_since_last_update",
    "competition_level",
    "main_intent",
    "content_type"
]

X = df[features].copy()

# Fill missing numeric values
numeric_cols = X.select_dtypes(include=["number"]).columns
X[numeric_cols] = X[numeric_cols].fillna(X[numeric_cols].median())

# Fill missing categorical values
categorical_cols = X.select_dtypes(include=["object"]).columns
X[categorical_cols] = X[categorical_cols].fillna("Unknown")

# One-hot encoding
X = pd.get_dummies(X, columns=categorical_cols)

print("Feature Vector Shape:", X.shape)
display(X.head())
```

## 3. The leakage hunt

*Attack your own features: label-derived columns, future windows, product flags. Show the test.*

```python
## Leakage Hunt

'''I reviewed all candidate features for possible leakage.

The goal was to identify:

- label-derived fields,
- future-window measurements,
- product-generated flags,
- recommendation outputs.

Only features available before the recommendation point were included.

The inspection did not identify any explicit target variable used in the feature vector.'''
```

```python
potential_leakage = [
    col for col in df.columns
    if "target" in col.lower()
    or "label" in col.lower()
    or "future" in col.lower()
    or "prediction" in col.lower()
    or "recommend" in col.lower()
]

print("Potential Leakage Columns:")
print(potential_leakage)
```
