## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

```python
## Ranked Actions

'''The ranked queue prioritizes pages that appear to have search opportunity based on observed search performance signals.

Pages with high impressions but relatively low click performance receive higher priority scores.

### Action: SEO_OPPORTUNITY

Reason Codes:

• HIGH_IMPRESSIONS_LOW_CTR
  - The page receives visibility but attracts relatively few clicks.

• HIGH_VISIBILITY_UNDERPERFORMING
  - The page appears in search results frequently but generates limited traffic.

• POSITION_WITH_OPPORTUNITY
  - The page has measurable visibility and may benefit from optimization.

The ranked queue should be used to prioritize review rather than automatically make content changes.'''
```

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

```python
## Intended Use

'''This playbook is intended for SEO Analysts and Content Managers.

The goal is to prioritize pages that may represent search opportunities based on historical search performance data.

The recommendations are intended as decision-support rather than automated decisions.

## Limits

The recommendations are based on observed historical data.

The model cannot determine:

- content quality
- future algorithm updates
- competitor activity
- business value
- user intent changes

The ranking should be treated as a directional signal rather than certainty.'''
```

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

```python
## Human Review Requirements

'''Before acting on a recommendation, a reviewer should verify:

- page relevance
- business importance
- content quality
- search intent alignment
- recent content updates

Human review remains necessary because search performance can be influenced by factors not included in the model.

## No-Go List

The following actions should never be automated:

- publishing content changes
- deleting content
- redirecting pages
- changing site structure
- making business decisions solely from model output

The recommendations should support human judgement rather than replace it.'''
```

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

```python
## Monitoring

'''The recommendation quality should be reviewed regularly.

Potential monitoring indicators include:

- declining CTR patterns
- decreasing impressions
- lower ranking stability
- reduced recommendation usefulness

## Retrain Triggers

The model should be reviewed or retrained when:

- new months of data become available
- feature distributions change substantially
- recommendation quality declines
- search behaviour changes significantly

Retraining helps ensure the model remains aligned with current search patterns.'''
```
